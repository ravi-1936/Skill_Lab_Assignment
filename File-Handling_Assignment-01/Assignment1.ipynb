Place `document.txt`, `data.txt`, `reverse.txt` and `stats.txt` in the same folder as this notebook.

## 1. Counting Words in a File

```python
def count_words(filename="document.txt"):
    with open(filename, 'r') as f:
        return sum(len(line.split()) for line in f)

print(count_words())
```

## 2. Finding and Replacing Text

```python
def replace_word(old_word, new_word, filename="data.txt"):
    with open(filename, 'r') as f:
        text = f.read()
    text = text.replace(old_word, new_word)
    with open(filename, 'w') as f:
        f.write(text)

replace_word("Hi", "Hello")
print(open("data.txt").read())
```

## 3. Reading a File in Reverse

```python
def reverse_file(filename="reverse.txt"):
    with open(filename, 'r') as f:
        lines = f.readlines()
    for line in reversed(lines):
        print(line.strip())

reverse_file()
```

## 4. Counting Lines, Words, and Characters

```python
def file_stats(filename="stats.txt"):
    with open(filename, 'r') as f:
        text = f.read()
    return len(text.splitlines()), len(text.split()), len(text)

lines, words, chars = file_stats()
print("Lines:", lines)
print("Words:", words)
print("Characters:", chars)
```
